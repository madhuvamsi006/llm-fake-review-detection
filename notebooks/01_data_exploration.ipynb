# Notebook 1: Data Exploration
Loads the Salminen et al. (2022) dataset and produces descriptive statistics and visualisations.

```python
import sys
sys.path.insert(0, '..')
from src.data.downloader import download_salminen_dataset
from src.data.loader import load_salminen
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

download_salminen_dataset()
df = load_salminen()
print(df.shape)
df.head()
```

```python
# Label distribution
df['label'].value_counts().plot(kind='bar', title='Label Distribution (0=Genuine, 1=GPT-2 Fake)')
plt.tight_layout()
plt.savefig('../results/figures/label_distribution.png', dpi=150)
plt.show()
```

```python
# Review length distribution
df['length'] = df['text'].str.split().str.len()
df.groupby('label')['length'].hist(bins=50, alpha=0.6)
plt.xlabel('Word count')
plt.title('Review Length by Label')
plt.savefig('../results/figures/length_distribution.png', dpi=150)
plt.show()
```

```python
# Category distribution
df['category'].value_counts().head(20).plot(kind='barh', figsize=(8,6))
plt.title('Top 20 Product Categories')
plt.tight_layout()
plt.savefig('../results/figures/category_distribution.png', dpi=150)
plt.show()
```
