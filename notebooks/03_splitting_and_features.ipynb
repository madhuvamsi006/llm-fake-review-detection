# Notebook 3: Leakage-Controlled Splitting & Feature Extraction
Implements product- and generator-aware splits, then extracts classical and transformer features.

```python
import sys
sys.path.insert(0, '..')
from src.data.loader import load_all
from src.data.splitter import product_aware_split, cross_generator_split

df = load_all()
print(df['generator'].value_counts())
```

```python
# In-distribution split
train, val, test = product_aware_split(df)
print(f'Train: {len(train)} | Val: {len(val)} | Test: {len(test)}')
```

```python
# Cross-generator split (hold out Claude)
train_cg, val_cg, test_id, test_cg = cross_generator_split(df, held_out_generator='claude')
print(f'Cross-gen test (Claude): {len(test_cg)}')
```

```python
# Save splits
train.to_csv('../data/processed/train.csv', index=False)
val.to_csv('../data/processed/val.csv', index=False)
test.to_csv('../data/processed/test_in_dist.csv', index=False)
test_cg.to_csv('../data/processed/test_cross_gen.csv', index=False)
print('Splits saved.')
```

```python
# Classical features
from src.features.classical import fit_transform_classical
features = fit_transform_classical(
    train['text'].tolist(),
    val['text'].tolist(),
    test['text'].tolist(),
)
print(f'Feature shape (train): {features["train"].shape}')
```
